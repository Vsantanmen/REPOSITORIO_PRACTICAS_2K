8- Realiza una función que devuelva el cuadrado de una lista de números enteros.
Ejemplos: Lista1 = [1,2,3,4]
          devuelve : Lista1* = [1,4,9,16]
          Utilizar la instrucción "map" para poder aplicar la función a un objeto iterable.

```python
def cuadrado_lista(lista):
    return list(map(lambda x: x ** 2, lista))

lista_original = [1, 2, 3, 4]
print(cuadrado_lista(lista_original))
```

9- Realiza una función que combine las siguientes dos colecciones iterables:
asignaturas = ['Matemáticas', 'Física', 'Química', 'Economía']
notas = [6.0, 3.5, 7.5, 8.0]

```python
def combinar_colecciones(lista1, lista2):
    return list(zip(lista1, lista2))

asignaturas = ['Matemáticas', 'Física', 'Química', 'Economía']
notas = [6.0, 3.5, 7.5, 8.0]
print(combinar_colecciones(asignaturas, notas))
```

10- Realiza una funcion que combine solo las 3 primeras posriciones de las mismas colecciones iterables

```python
def combinar_primeras_posiciones(lista1, lista2):
    return list(zip(lista1[:3], lista2[:3]))

asignaturas = ['Matemáticas', 'Física', 'Química', 'Economía']
notas = [6.0, 3.5, 7.5, 8.0]
print(combinar_primeras_posiciones(asignaturas, notas))
```
