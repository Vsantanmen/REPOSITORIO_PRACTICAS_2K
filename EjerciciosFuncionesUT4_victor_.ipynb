3. Escribir una función que reciba un número entero positivo y devuelva su factorial.

```python
def factorial(n):
  if n < 0:
    raise ValueError("El factorial no está definido para números negativos")
  elif n == 0:
    return 1
  else:
    resultado = 1
    for i in range(1, n + 1):
      resultado *= i
    return resultado

print(factorial(5))
```

5. Escribir una función que calcule el área de un círculo y otra que calcule el volumen de un cilindro usando la primera función.

```python
def area_circulo(radio):
    pi = 3.141592653589793
    return pi * (radio ** 2)

def volumen_cilindro(radio, altura):
    return area_circulo(radio) * altura


print(f"Área del círculo: {area_circulo(3)}")
print(f"Volumen del cilindro: {volumen_cilindro(3, 5)}")
```

6. Escribir una función que reciba una muestra de números en una lista y devuelva su media.

```python
def calcular_media(lista_numeros):
    if len(lista_numeros) == 0:
        return 0
    return sum(lista_numeros) / len(lista_numeros)


muestra = [10, 20, 30, 40, 50]
print(f"La media es: {calcular_media(muestra)}")
```

7. Escribir una función que reciba una muestra de números en una lista y devuelva otra lista con sus cuadrados.

```python
def calcular_cuadrados(lista_numeros):
    cuadrados = []
    for numero in lista_numeros:
        cuadrados.append(numero ** 2)
    return cuadrados


muestra = [1, 2, 3, 4, 5]
print(f"Lista de cuadrados: {calcular_cuadrados(muestra)}")
```

8. Escribir una función que reciba una muestra de números en una lista y devuelva un diccionario con su media, varianza y desviación típica.

```python
def calcular_estadisticas(lista_numeros):
    n = len(lista_numeros)
    if n == 0:
        return {"media": 0, "varianza": 0, "desviacion_tipica": 0}

    media = sum(lista_numeros) / n

    suma_cuadrados_diferencias = 0
    for numero in lista_numeros:
        suma_cuadrados_diferencias += (numero - media) ** 2

    varianza = suma_cuadrados_diferencias / n
    desviacion_tipica = varianza ** 0.5

    return {
        "media": media,
        "varianza": varianza,
        "desviacion_tipica": desviacion_tipica
    }


muestra = [2, 4, 4, 4, 5, 5, 7, 9]
print(calcular_estadisticas(muestra))
```

9. Escribir una función que calcule el máximo común divisor de dos números y otra que calcule el mínimo común múltiplo.

```python
def maximo_comun_divisor(a, b):
    while b != 0:
        a, b = b, a % b
    return a

def minimo_comun_multiplo(a, b):
    if a == 0 or b == 0:
        return 0
    return abs(a * b) // maximo_comun_divisor(a, b)


num1 = 24
num2 = 36

print(f"El M.C.D. de {num1} y {num2} es: {maximo_comun_divisor(num1, num2)}")
print(f"El m.c.m. de {num1} y {num2} es: {minimo_comun_multiplo(num1, num2)}")
```

10. Escribir una función que convierta un número decimal en binario y otra que convierta un número binario en decimal.

```python
def decimal_a_binario(numero_decimal):
    if numero_decimal == 0:
        return "0"

    binario = ""
    while numero_decimal > 0:
        residuo = numero_decimal % 2
        binario = str(residuo) + binario
        numero_decimal = numero_decimal // 2
    return binario

def binario_a_decimal(numero_binario):
    numero_binario = str(numero_binario)
    decimal = 0
    posicion = 0

    for digito in reversed(numero_binario):
        if digito == "1":
            decimal += 2 ** posicion
        posicion += 1
    return decimal


n_decimal = 26
n_binario = "11010"

print(f"El número {n_decimal} en binario es: {decimal_a_binario(n_decimal)}")
print(f"El binario {n_binario} en decimal es: {binario_a_decimal(n_binario)}")
```

11. Escribir un programa que reciba una cadena de caracteres y devuelva un diccionario con cada palabra que contiene y su frecuencia. Escribir otra función que reciba el diccionario generado con la función anterior y devuelva una tupla con la palabra más repetida y su frecuencia.

```python
def contar_palabras(cadena):
    palabras = cadena.lower().split()
    frecuencias = {}

    for palabra in palabras:
        palabra = palabra.strip(",.!?;:()")
        if palabra:
            if palabra in frecuencias:
                frecuencias[palabra] += 1
            else:
                frecuencias[palabra] = 1

    return frecuencias

def palabra_mas_repetida(diccionario_frecuencias):
    if not diccionario_frecuencias:
        return ("", 0)

    palabra_max = ""
    frecuencia_max = -1

    for palabra, frecuencia in diccionario_frecuencias.items():
        if frecuencia > frecuencia_max:
            frecuencia_max = frecuencia
            palabra_max = palabra

    return (palabra_max, frecuencia_max)


texto = "Hola mundo, hola de nuevo mundo. Hola otra vez."
resultado_diccionario = contar_palabras(texto)
resultado_tupla = palabra_mas_repetida(resultado_diccionario)

print(resultado_diccionario)
print(resultado_tupla)
```
